# 📊 Blog 18: Ranking Anomalies with Smart Alert Prioritization


In this notebook, we explore a heuristic-based approach to prioritize anomaly alerts based on **severity**, **frequency**, and **impact**.

We simulate alert data and generate:
- A heatmap of anomaly severity
- A bar chart of top alert scores
- A cumulative impact curve over time

This approach helps address *alert fatigue* by ranking which alerts need the most urgent attention.


```

import pandas as pd
import numpy as np

# Simulate anomaly data
np.random.seed(42)
n = 100
timestamps = pd.date_range("2024-01-01", periods=n, freq="H")
severity = np.random.rand(n)
frequency = np.random.randint(1, 10, size=n)
impact = np.random.rand(n) * 100

alert_score = severity * frequency * impact

df = pd.DataFrame({
    "timestamp": timestamps,
    "severity": severity,
    "frequency": frequency,
    "impact": impact,
    "alert_score": alert_score
})

df.head()
```

### 🔥 Heatmap: Anomaly Severity by Hour

```

import seaborn as sns
import matplotlib.pyplot as plt

heatmap_data = df.pivot_table(index=df["timestamp"].dt.date, 
                              columns=df["timestamp"].dt.hour, 
                              values="severity", aggfunc='mean')

plt.figure(figsize=(10, 6))
sns.heatmap(heatmap_data, cmap="Reds", linewidths=0.1)
plt.title("Anomaly Severity Heatmap")
plt.show()
```

### 🚨 Bar Chart: Top 15 Alerts by Score

```

top_alerts = df.sort_values("alert_score", ascending=False).head(15)

plt.figure(figsize=(10, 6))
sns.barplot(x="alert_score", y="timestamp", data=top_alerts, palette="viridis")
plt.title("Top 15 Alert Scores by Timestamp")
plt.xlabel("Alert Score")
plt.ylabel("Timestamp")
plt.tight_layout()
plt.show()
```

### 📈 Cumulative Anomaly Impact Over Time

```

df_sorted = df.sort_values("timestamp")
df_sorted["cumulative_impact"] = df_sorted["impact"].cumsum()

plt.figure(figsize=(10, 6))
plt.plot(df_sorted["timestamp"], df_sorted["cumulative_impact"], label="Cumulative Impact")
plt.xlabel("Timestamp")
plt.ylabel("Cumulative Impact")
plt.title("Cumulative Anomaly Impact Over Time")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

## ✅ Summary


By scoring each alert using a simple formula `severity × frequency × impact`, we can:

- Visualize hotspots via heatmaps
- Prioritize actions based on top alert scores
- Track how much potential damage accumulates over time

This approach is lightweight and customizable, ideal for initial deployments before ML-based alert systems.
